```python
import requests
from datetime import date
```

```python
params = {
    "lang": "fr",
    "ordre": "antechronologique",
    "transposition": "donneescolonne",
    "anneeDebut": "2010",
    "anneeFin": str(date.today().year),
    "revision": "sansrevisions",
}
INSEE_URL = f"https://bdm.insee.fr/series/001565530/csv"
response = requests.get(INSEE_URL, params=params)
```

```python
response.content
```

```python
# Save the real INSEE response as success.zip for tests
from pathlib import Path

# Ensure the directory exists
zip_path = Path("economic_overview/tests/mock_web_data/insee/data/success.zip")
zip_path.parent.mkdir(parents=True, exist_ok=True)

# Save the response content (which is a ZIP file) to the test file
with open(zip_path, "wb") as f:
    f.write(response.content)

print(f"Saved INSEE response to {zip_path}")
print(f"File size: {len(response.content)} bytes")
```
